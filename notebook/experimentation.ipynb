```python
import os
import matplotlib.pyplot as plt
from PIL import Image
import numpy as np
```

### Cell 2 — count images per class

```python
dataset_path = "../dataset"
class_names = os.listdir(dataset_path)
class_names = [c for c in class_names if os.path.isdir(os.path.join(dataset_path, c))]
class_names.sort()

print("Classes found:", class_names)
print()

total = 0
for cls in class_names:
    cls_path = os.path.join(dataset_path, cls)
    count = len(os.listdir(cls_path))
    total += count
    print(f"  {cls:20s}: {count:5d} images")

print(f"\n  {'TOTAL':20s}: {total:5d} images")
```

###Cell 3 — visualise sample images from each class

```python

fig, axes = plt.subplots(1, 5, figsize=(20, 4))
fig.suptitle("Sample Retinal Images by Severity", fontsize=14)

for i, cls in enumerate(class_names):
    cls_path = os.path.join(dataset_path, cls)
    img_file = os.listdir(cls_path)[0]         # grab first image
    img = Image.open(os.path.join(cls_path, img_file))
    axes[i].imshow(img)
    axes[i].set_title(cls, fontsize=10)
    axes[i].axis("off")

plt.tight_layout()
plt.savefig("../static/sample_grid.png", dpi=100)
plt.show()
print("Saved sample grid to static/")
```
